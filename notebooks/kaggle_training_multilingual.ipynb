# 🌐 VulnLocator V2.0: Multilingual Generalist Training on Kaggle Dual T4 GPUs

This notebook fine-tunes **CodeBERT** (`microsoft/codebert-base`) for multilingual vulnerability detection on the **CrossVul** dataset across **Java, Python, PHP, JavaScript, Go, and Ruby**.

### 🛡️ Why Specialist vs. Generalist (Mixture of Experts)?
When a single model is trained simultaneously on systems C/C++ memory corruption and high-level web vulnerabilities (SQLi, XSS, eval injection), **Catastrophic Forgetting** causes pointer and buffer overflow detection to degrade significantly.

**VulnLocator V2.0** solves this with a Router Architecture:
- **Specialist Expert:** C/C++ memory corruption (PrimeVul)
- **Generalist Expert:** Multilingual web/API vulnerabilities (CrossVul)
- **Router:** AST/Syntactic language detection directs inputs to the proper expert.

```python
import os
import time
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_recall_fscore_support,
    roc_auc_score,
    classification_report,
    confusion_matrix,
)
from tqdm.auto import tqdm

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device_count = torch.cuda.device_count()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device} | Available GPUs: {device_count}")
for i in range(device_count):
    print(f"  [GPU {i}] {torch.cuda.get_device_name(i)}")
```

## 1. Load Multilingual CrossVul Dataset

```python
df = pd.read_csv("data/multilingual_sample.csv")
df = df.dropna(subset=["func", "target"])
df["target"] = df["target"].astype(int)
print(f"Total multilingual samples: {len(df):,}")
print("\nSamples per language:")
print(df["language"].value_counts())
print("\nClass balance:")
print(df["target"].value_counts(normalize=True).map(lambda n: f"{n:.2%}"))
```

## 2. In-Memory Pre-Tokenization Pipeline (Zero CPU Bottleneck)

```python
MODEL_NAME = "microsoft/codebert-base"
MAX_LEN = 256
BATCH_SIZE = 32
EPOCHS = 3
LR = 2e-5

train_df, temp_df = train_test_split(df, test_size=0.3, stratify=df["target"], random_state=SEED)
val_df, test_df = train_test_split(temp_df, test_size=0.5, stratify=temp_df["target"], random_state=SEED)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Pre-tokenizing multilingual splits into GPU-ready TensorDatasets...")
t0 = time.time()
train_enc = tokenizer(train_df["func"].astype(str).tolist(), truncation=True, padding="max_length", max_length=MAX_LEN, return_tensors="pt")
val_enc = tokenizer(val_df["func"].astype(str).tolist(), truncation=True, padding="max_length", max_length=MAX_LEN, return_tensors="pt")
test_enc = tokenizer(test_df["func"].astype(str).tolist(), truncation=True, padding="max_length", max_length=MAX_LEN, return_tensors="pt")
print(f"Pre-tokenization done in {time.time() - t0:.2f}s")

train_loader = DataLoader(TensorDataset(train_enc["input_ids"], train_enc["attention_mask"], torch.tensor(train_df["target"].values)), batch_size=BATCH_SIZE, shuffle=True, pin_memory=True)
val_loader = DataLoader(TensorDataset(val_enc["input_ids"], val_enc["attention_mask"], torch.tensor(val_df["target"].values)), batch_size=BATCH_SIZE, shuffle=False, pin_memory=True)
test_loader = DataLoader(TensorDataset(test_enc["input_ids"], test_enc["attention_mask"], torch.tensor(test_df["target"].values)), batch_size=BATCH_SIZE, shuffle=False, pin_memory=True)
```

## 3. Initialize Model with Multi-GPU DataParallel & Weighted Loss

```python
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=2)
model.to(device)

if device_count > 1:
    print(f"Enabling torch.nn.DataParallel across {device_count} GPUs!")
    model = nn.DataParallel(model)

counts = train_df["target"].value_counts().sort_index()
weights = torch.tensor([1.0 / counts[0], 1.0 / counts[1]], dtype=torch.float32)
weights = weights / weights.sum()
loss_fn = nn.CrossEntropyLoss(weight=weights.to(device))

optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=0.01)
scaler = torch.amp.GradScaler("cuda")
```

## 4. Multi-GPU Training & Evaluation Loop

```python
def evaluate(loader, sweep_threshold=False):
    model.eval()
    all_labels, all_probs = [], []
    with torch.no_grad():
        for batch in loader:
            ids, mask, labels = batch[0].to(device), batch[1].to(device), batch[2].to(device)
            with torch.amp.autocast("cuda"):
                outputs = model(input_ids=ids, attention_mask=mask)
                probs = torch.softmax(outputs.logits, dim=1)[:, 1]
            all_labels.extend(labels.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())
    
    all_labels, all_probs = np.array(all_labels), np.array(all_probs)
    auc = roc_auc_score(all_labels, all_probs)
    best_thresh, best_f1 = 0.5, 0.0
    if sweep_threshold:
        for thresh in np.arange(0.05, 0.95, 0.02):
            preds = (all_probs >= thresh).astype(int)
            _, _, f1, _ = precision_recall_fscore_support(all_labels, preds, average="binary", zero_division=0)
            if f1 > best_f1:
                best_f1, best_thresh = f1, float(thresh)
    preds = (all_probs >= best_thresh).astype(int)
    p, r, f1, _ = precision_recall_fscore_support(all_labels, preds, average="binary", zero_division=0)
    return {"precision": p, "recall": r, "f1": f1, "auc": auc, "threshold": best_thresh}, all_labels, preds, all_probs

best_val_auc = 0.0
best_val_thresh = 0.5

for epoch in range(1, EPOCHS + 1):
    model.train()
    total_loss, t_start = 0.0, time.time()
    pbar = tqdm(train_loader, desc=f"Epoch {epoch}/{EPOCHS}")
    for batch in pbar:
        ids, mask, labels = batch[0].to(device), batch[1].to(device), batch[2].to(device)
        optimizer.zero_grad()
        with torch.amp.autocast("cuda"):
            outputs = model(input_ids=ids, attention_mask=mask)
            loss = loss_fn(outputs.logits, labels)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        total_loss += loss.item()
        pbar.set_postfix({"loss": f"{loss.item():.4f}"})
    
    elapsed = time.time() - t_start
    val_m, _, _, _ = evaluate(val_loader, sweep_threshold=True)
    print(f"Epoch {epoch}/{EPOCHS} [{elapsed:.1f}s] - Loss: {total_loss/len(train_loader):.4f} | Val AUC: {val_m['auc']:.4f} | F1: {val_m['f1']:.3f} | Thresh: {val_m['threshold']:.2f}")
    if val_m["auc"] > best_val_auc:
        best_val_auc, best_val_thresh = val_m["auc"], val_m["threshold"]
```

## 5. Final Multilingual Test Evaluation

```python
test_m, test_labels, _, test_probs = evaluate(test_loader, sweep_threshold=False)
test_preds = (test_probs >= best_val_thresh).astype(int)

print(f"Final Test ROC-AUC: {test_m['auc']:.4f}")
print(f"Decision Threshold: {best_val_thresh:.2f}")
print("\nClassification Report:")
print(classification_report(test_labels, test_preds, target_names=["Safe", "Vulnerable"], digits=4))
print("Confusion Matrix:")
print(confusion_matrix(test_labels, test_preds))
```

## 6. Export Multilingual Generalist Artifacts

```python
OUTPUT_DIR = "models/codebert-multilingual"
os.makedirs(OUTPUT_DIR, exist_ok=True)
model_to_save = model.module if hasattr(model, "module") else model
model_to_save.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print(f"Saved Multilingual Generalist weights to {OUTPUT_DIR}")
```
